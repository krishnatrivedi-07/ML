```python
df = pd.read_csv("Wine.csv")
df.head()
```

```python
# Cell 3 - Prepare Data
X = df.drop('Customer_Segment', axis=1)  # All features, except target class
y = df['Customer_Segment']  # Target - type of wine
```

```python
# Cell 4 - Standardize + PCA
scaler = StandardScaler()  # For scaling data
X_scaled = scaler.fit_transform(X)  # Mean=0, Std=1

pca = PCA(n_components=2)  # Reduce to 2 dimensions for visualization
X_pca = pca.fit_transform(X_scaled)  # Transform data



```

```python
# Cell 5 - Visualization
plt.figure(figsize=(7,5))
plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA - Wine Dataset (2 Components)')
plt.colorbar(label='Wine Class')
plt.show()
```

```python
# Check variance
print("Variance captured:", pca.explained_variance_ratio_)
print("Total Variance:", sum(pca.explained_variance_ratio_))
```
